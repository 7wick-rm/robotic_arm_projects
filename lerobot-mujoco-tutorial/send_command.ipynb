# Send commands to the running simulation

Keep **deploy_policy.ipynb** running in its own tab (its loop cell must be executing), then use this notebook.
Same port as the deploy notebook (`5005`). Your gesture teammate can send the same text from their program.

```python
import socket
PORT = 5005
def send(text, host='127.0.0.1', port=PORT):
    socket.socket(socket.AF_INET, socket.SOCK_DGRAM).sendto(text.encode(), (host, port))
    print("sent:", text)
```

### Option A: buttons (needs `ipywidgets`, skip if the next cell prints a message instead)

```python
try:
    import ipywidgets as w
    from IPython.display import display
    buttons = []
    for name in ['wrench', 'screwdriver', 'hammer', 'pliers', 'reset', 'quit']:
        b = w.Button(description=name)
        b.on_click(lambda _b, n=name: send(n))
        buttons.append(b)
    display(w.HBox(buttons))
except ImportError:
    print("ipywidgets not installed -> use Option B below (or: pip install ipywidgets)")
```
