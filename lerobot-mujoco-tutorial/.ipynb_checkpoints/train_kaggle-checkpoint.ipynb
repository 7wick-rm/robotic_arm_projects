# Train SmolVLA on the garage dataset  (Kaggle, free GPU)

**Before you start (once)**
1. Put your dataset on the Hugging Face Hub as a **private dataset**. On your laptop run `python upload_dataset_to_hf.py --repo YOUR_USERNAME/garage_pnp` (the same script as for Colab).
2. Create a Hugging Face **write** token (huggingface.co/settings/tokens). In this notebook open **Add-ons -> Secrets**, add a secret named `HF_TOKEN`, and tick it for this notebook. Write access is needed because checkpoints are pushed to a private model repo.
3. In the right-hand **Session options** choose **Accelerator: GPU T4 x2** (see the note after the GPU check) and **Internet: On**. Kaggle asks you to verify your phone number the first time you turn Internet on.

**Two ways to run it**
- *Interactive:* run the cells top to bottom. Good for a short test.
- *Background (recommended for the long run):* **Save Version -> Save & Run All (Commit)**. It keeps running if you close the browser, and needs no clicks because the token comes from the Secret.

Kaggle limits GPU time per week (check your own quota on your profile) and each session to roughly 12 hours. So the training cell stops itself after `MAX_RUN_HOURS`, and the newest checkpoint is always pushed to your private model repo on the Hub. To continue in a new session, run the cells again but run the **Resume** cell instead of **Train**.

```python
import os, shutil, subprocess, threading, time
from pathlib import Path

# ----------------------------- SETTINGS -----------------------------
HF_DATASET    = 'YOUR_USERNAME/garage_pnp'            # <-- the private dataset repo you uploaded from your laptop
HF_CKPT_REPO  = 'YOUR_USERNAME/smolvla_garage_ckpt'   # <-- a private MODEL repo for checkpoints (created automatically)
KAGGLE_INPUT_DIR = None   # optional: if you uploaded the dataset as a Kaggle Dataset instead, the folder that holds
                          # data/ and meta/, e.g. '/kaggle/input/garage-pnp/demo_data_garage'
REPO_NAME     = 'garage_pnp'

STEPS         = 15_000     # 8-10k is already a decent first try
BATCH_SIZE    = 16         # 8 if you get CUDA out-of-memory
CHUNK         = 10         # chunk_size = n_action_steps (training and deployment must match)
SAVE_FREQ     = 2_500
NUM_WORKERS   = 3          # Kaggle has 4 CPU cores
MAX_RUN_HOURS = 10.5       # the training cell stops itself after this, so the session limit never cuts it off
# --------------------------------------------------------------------
assert 'YOUR_USERNAME' not in HF_DATASET + HF_CKPT_REPO, "set HF_DATASET and HF_CKPT_REPO to your own Hugging Face names"

# Big files go on the large temporary disk. /kaggle/working is capped at 20 GB and is saved as notebook output,
# so we only put the final model there.
BIG       = '/kaggle/temp' if os.path.isdir('/kaggle/temp') else '/tmp'
WORK      = f'{BIG}/work'
DATA_ROOT = f'{BIG}/demo_data_garage'
LOCAL_OUT = f'{BIG}/ckpt/smolvla_garage'
FINAL_OUT = '/kaggle/working/smolvla_garage_final'
for d in (WORK, f'{BIG}/hf_home', f'{BIG}/hf_datasets_cache'):
    os.makedirs(d, exist_ok=True)

# model downloads and the `datasets` cache (a second copy of the data!) also go to the big disk
os.environ['HF_HOME'] = f'{BIG}/hf_home'
os.environ['HF_DATASETS_CACHE'] = f'{BIG}/hf_datasets_cache'
try:
    import hf_transfer                       # faster downloads when available
    os.environ['HF_HUB_ENABLE_HF_TRANSFER'] = '1'
except ImportError:
    pass
print("big disk:", BIG, "| free GB:", round(shutil.disk_usage(BIG).free / 1e9, 1))
```

```python
# Hugging Face login: token from the Kaggle Secret named HF_TOKEN
try:
    from kaggle_secrets import UserSecretsClient
    HF_TOKEN = UserSecretsClient().get_secret('HF_TOKEN')
except Exception as exc:
    print("No Kaggle secret HF_TOKEN found:", type(exc).__name__)
    HF_TOKEN = None
if not HF_TOKEN:
    from getpass import getpass
    HF_TOKEN = getpass('Paste your Hugging Face WRITE token (interactive sessions only): ')
os.environ['HF_TOKEN'] = HF_TOKEN            # train_model.py downloads the base model with this

from huggingface_hub import HfApi, snapshot_download, login
login(token=HF_TOKEN)
api = HfApi(token=HF_TOKEN)
print("logged in as:", api.whoami()['name'])
```

```python
# dataset: get it onto the big local disk (from the Hub, or from a Kaggle Dataset you attached)
def free_gb(path):
    return shutil.disk_usage(path).free / 1e9

if not os.path.exists(f'{DATA_ROOT}/meta/info.json'):
    if KAGGLE_INPUT_DIR:
        size_gb = sum(f.stat().st_size for f in Path(KAGGLE_INPUT_DIR).rglob('*') if f.is_file()) / 1e9
    else:
        size_gb = sum((s.size or 0) for s in api.dataset_info(HF_DATASET, files_metadata=True).siblings) / 1e9
    need_gb = 2.2 * size_gb + 10        # data + the `datasets` cache copy + model weights and checkpoints
    print(f"dataset: {size_gb:.1f} GB | free on {BIG}: {free_gb(BIG):.1f} GB | needed: about {need_gb:.0f} GB")
    assert free_gb(BIG) > need_gb, "Not enough free disk for the dataset plus its cache. See the note about shrinking the images."

    if KAGGLE_INPUT_DIR:
        shutil.copytree(KAGGLE_INPUT_DIR, DATA_ROOT)       # /kaggle/input is read-only, so work on a copy
    else:
        snapshot_download(repo_id=HF_DATASET, repo_type='dataset', local_dir=DATA_ROOT, max_workers=8, token=HF_TOKEN)
assert os.path.exists(f'{DATA_ROOT}/meta/info.json'), "meta/info.json is missing - data/ and meta/ must be at the top level of the dataset"

# training script (identical to the one in the tutorial repo)
!wget -q -O {WORK}/train_model.py https://raw.githubusercontent.com/jeongeun980906/lerobot-mujoco-tutorial/master/train_model.py
print(os.listdir(WORK))
!df -h {BIG}
```

```python
# sanity-check the dataset before spending GPU time
import json
info = json.load(open(f'{DATA_ROOT}/meta/info.json'))
print("episodes:", info['total_episodes'], " frames:", info['total_frames'], " fps:", info['fps'])
for k, v in info['features'].items():
    print(f"  {k:28s} {v['dtype']:8s} {v['shape']}")
print("tasks:")
for line in open(f'{DATA_ROOT}/meta/tasks.jsonl'):
    print("  ", json.loads(line))
assert info['features']['observation.state']['shape'] == [8] and info['features']['action']['shape'] == [8]
```

```python
yaml_text = f'''
dataset:
  repo_id: {REPO_NAME}
  root: {DATA_ROOT}
policy:
  type: smolvla
  chunk_size: {CHUNK}
  n_action_steps: {CHUNK}
  device: cuda
  use_amp: false            # T4 has no bf16
output_dir: {LOCAL_OUT}
batch_size: {BATCH_SIZE}
job_name: smolvla_garage
resume: false
seed: 42
num_workers: {NUM_WORKERS}
steps: {STEPS}
eval_freq: -1
log_freq: 50
save_checkpoint: true
save_freq: {SAVE_FREQ}
use_policy_training_preset: true
wandb:
  enable: false
'''
YAML = f'{WORK}/smolvla_garage.yaml'
open(YAML, 'w').write(yaml_text)
print(yaml_text)
```

### Checkpoint sync and the training runner
Checkpoints are written to the big local disk. Every two minutes a background thread pushes the newest finished one to your private model repo on the Hub and removes the older ones there (each checkpoint is several GB), so a session that ends unexpectedly costs you at most `SAVE_FREQ` steps.

```python
api.create_repo(HF_CKPT_REPO, repo_type='model', private=True, exist_ok=True)

def latest_local_ckpt():
    last = Path(LOCAL_OUT) / 'checkpoints' / 'last'
    return Path(os.path.realpath(last)) if last.exists() else None     # 'last' is only updated AFTER a save finished

def remote_steps():
    names = set()
    for f in api.list_repo_files(HF_CKPT_REPO, repo_type='model'):
        parts = f.split('/')
        if len(parts) > 2 and parts[0] == 'checkpoints':
            names.add(parts[1])
    return sorted(names)

def sync_to_hub():
    src = latest_local_ckpt()
    if src is None or src.name in remote_steps():
        return
    api.upload_folder(folder_path=str(src), path_in_repo=f'checkpoints/{src.name}',
                      repo_id=HF_CKPT_REPO, repo_type='model', commit_message=f'checkpoint {src.name}')
    for old in remote_steps():
        if old != src.name:
            api.delete_folder(path_in_repo=f'checkpoints/{old}', repo_id=HF_CKPT_REPO, repo_type='model')
    try:
        api.super_squash_history(repo_id=HF_CKPT_REPO, repo_type='model')   # drop old versions so they stop using storage
    except Exception as exc:
        print("[sync] (history squash skipped:", exc, ")")
    print(f"[sync] checkpoint {src.name} saved to {HF_CKPT_REPO}")

def run_training(train_args):
    """Runs train_model.py, streams its output, syncs checkpoints, and stops itself after MAX_RUN_HOURS."""
    stop = threading.Event()

    def syncer():
        while not stop.is_set():
            try:
                sync_to_hub()
            except Exception as exc:
                print("[sync] error:", exc)
            stop.wait(120)

    th = threading.Thread(target=syncer, daemon=True)
    th.start()
    deadline = time.time() + MAX_RUN_HOURS * 3600
    p = subprocess.Popen(['python', 'train_model.py'] + train_args, cwd=WORK, env={**os.environ, 'PYTHONUNBUFFERED': '1'},
                         stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    timed_out = False
    for line in p.stdout:
        print(line, end='')
        if time.time() > deadline:
            print(f"\nReached MAX_RUN_HOURS={MAX_RUN_HOURS}. Stopping; the newest checkpoint is kept.")
            p.terminate()
            timed_out = True
            break
    p.wait()
    stop.set()
    th.join()
    try:
        sync_to_hub()
    except Exception as exc:
        print("[sync] final sync error:", exc)
    print("training process exited with code", p.returncode)
    if timed_out:
        print("Training is NOT finished. Start a new session, run the cells up to the yaml cell and the sync cell, then run the Resume cell.")
    return p.returncode
```

### Train  (output is streamed below)
Run this cell for a fresh start. If a session already trained part of the way, use the **Resume** cell below instead.

```python
run_training(['--config_path', YAML])
```

### [Only if a session ended before training finished] Resume
In the new session run the cells **install, settings, login, dataset, sanity check, yaml, sync helpers**, then this cell. It downloads the newest checkpoint from your model repo and continues from that step.

```python
def restore_from_hub():
    steps = remote_steps()
    assert steps, f"no checkpoint found in {HF_CKPT_REPO} yet"
    name = steps[-1]
    snapshot_download(repo_id=HF_CKPT_REPO, repo_type='model', allow_patterns=[f'checkpoints/{name}/**'],
                      local_dir=LOCAL_OUT, token=HF_TOKEN)
    root = Path(LOCAL_OUT) / 'checkpoints'
    last = root / 'last'
    if last.is_symlink() or last.exists():
        last.unlink()
    os.symlink(name, last)
    print("restored checkpoint", name)
    return root / 'last' / 'pretrained_model' / 'train_config.json'

cfg_json = restore_from_hub()
run_training([f'--config_path={cfg_json}', '--resume=true'])
```

### Finished: where is my model?
The final model is copied to `/kaggle/working/smolvla_garage_final`, which Kaggle keeps as notebook output (after **Save & Run All**, download it from the notebook's *Output* tab). The same checkpoint is also in your private model repo on the Hub.

```python
src = latest_local_ckpt()
assert src is not None, "no finished checkpoint found on the local disk"
shutil.rmtree(FINAL_OUT, ignore_errors=True)
shutil.copytree(src / 'pretrained_model', FINAL_OUT)
size_gb = sum(f.stat().st_size for f in Path(FINAL_OUT).rglob('*') if f.is_file()) / 1e9
print(f"final model: step {src.name}, {size_gb:.2f} GB -> {FINAL_OUT}")
print(os.listdir(FINAL_OUT))
print()
print("On your laptop, to get it from the Hub:")
print(f"  huggingface-cli download {HF_CKPT_REPO} --include 'checkpoints/{src.name}/pretrained_model/*' --local-dir ./ckpt/smolvla_garage")
print(f"then use  ./ckpt/smolvla_garage/checkpoints/{src.name}/pretrained_model  as the checkpoint folder.")
```
